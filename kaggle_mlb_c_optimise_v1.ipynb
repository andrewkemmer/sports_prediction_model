# MLB C arm: ablation, sweep, seeds, likelihood, starter-awareness

Five studies on the fitted run-model rating, all measured downstream on the moneyline:

1. **Ablation** — team one-hots alone, then each nuisance block, then the shipped design.
2. **Sweep** — 6 ridge penalties x 5 recency half-lives.
3. **Seeds** — the winning configuration refit under three member seeds, every arm refit.
4. **Likelihood** — Gaussian ridge vs Poisson GLM at the winning penalty.
5. **Starter-aware** — the team's own starting pitcher added to the design.

Arms A and B do not see the rating, so they are fitted once per seed and reused; each variant costs one arm-C walk forward.
