# NBA moneyline A/B/C: DARKO strength block vs fitted point rating

A incumbent 92 columns. B removes the eleven-column strength block (pace-scaled DARKO offence/defence blends, expected points, the pre-computed matchup net). C adds the seven columns from a causal ridge of points per 100 possessions on team-game rows: off/def rating home/away/diff plus net_rating_diff. Availability (star_out_*) stays in both arms so the comparison isolates strength.
