# NHL moneyline A/B/C: Elo + win% vs fitted goal rating

A incumbent, B removes the six Elo/win% names, C adds the seven columns from a causal ridge of goals per 60 minutes of team ice time on side rows, with a team-offence and a team-defence one-hot on the same row so each effect is estimated net of the opponents faced: off/def rating home/away/diff plus net_rating_diff. Replaces the recombination of xPM cells that arm B already serves.
